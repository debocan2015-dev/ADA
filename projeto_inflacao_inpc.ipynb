Análise do INPC: Do Governo Temer aos Dias Atuais
Este projeto tem como objetivo analisar o comportamento da inflação oficial da classe trabalhadora (INPC) desde o início do governo Temer (maio de 2016) até o período recente, utilizando a biblioteca Pandas para manipulação, limpeza e cruzamento de dados públicos extraídos da API do Banco Central do Brasil.:

Pergunta 1: Qual é o período total da série e o volume de dados analisados?
Objetivo: Identificar o escopo temporal e a quantidade de registros mensais disponíveis no DataFrame.

```python
total_meses = len(df_analise)
data_inicio = df_analise['data'].min().strftime('%m/%Y')
data_fim = df_analise['data'].max().strftime('%m/%Y')

print(f"Total de meses analisados: {total_meses} meses.")
print(f"Período compreendido: de {data_inicio} até {data_fim}.")
```

Pergunta 2: Quais foram os 5 meses com maior inflação (picos) no período?

```python
picos_inflacao = df_analise.nlargest(5, 'valor')
# Formatando a exibição da data para Mês/Ano
picos_inflacao['mes_ano'] = picos_inflacao['data'].dt.strftime('%m/%Y')
print("Os 5 meses com maior inflação registrada:")
print(picos_inflacao[['mes_ano', 'valor']])
```

Pergunta 3: Qual foi a inflação média mensal geral de todo o período?

```python
media_geral = df_analise['valor'].mean()
print(f"Inflação média mensal geral no período: {media_geral:.2f}%")
```

Pergunta 4: Qual foi a inflação acumulada ano a ano?

```python
df_analise['ano'] = df_analise['data'].dt.year
inflacao_anual = df_analise.groupby('ano')['valor'].sum().reset_index()
inflacao_anual.columns = ['Ano', 'Inflação Acumulada Simples (%)']
print(inflacao_anual)
```

Pergunta 5: Quantos meses registraram deflação (valores negativos de INPC)?

```python
deflacao = df_analise[df_analise['valor'] < 0]
print(f"Número de meses com deflação: {len(deflacao)}")
if len(deflacao) > 0:
    deflacao_copia = deflacao.copy()
    deflacao_copia['mes_ano'] = deflacao_copia['data'].dt.strftime('%m/%Y')
    print("\nMeses com deflação:")
    print(deflacao_copia[['mes_ano', 'valor']])
```

Pergunta 6: Qual foi o comportamento da inflação durante o recorte específico do Governo Temer (2016-2018)?

```python
governo_temer = df_analise[(df_analise['data'] >= '2016-05-01') & (df_analise['data'] <= '2018-12-31')]
soma_temer = governo_temer['valor'].sum()
media_temer = governo_temer['valor'].mean()

print(f"Período do Governo Temer analisado: {len(governo_temer)} meses")
print(f"Inflação acumulada (soma simples) no período: {soma_temer:.2f}%")
print(f"Média mensal de inflação na gestão: {media_temer:.2f}%")
```

Pergunta 7: Quais foram os meses em que a inflação mensal ultrapassou 1.0%?


```python
inflacao_alta = df_analise[df_analise['valor'] > 1.0].copy()
inflacao_alta['mes_ano'] = inflacao_alta['data'].dt.strftime('%m/%Y')

print(f"Total de meses com inflação acima de 1%: {len(inflacao_alta)}")
print("\nPrimeiros registros encontrados:")
print(inflacao_alta[['mes_ano', 'valor']].head())
```

Pergunta 8: Como classificar os meses de acordo com a intensidade da inflação?

```python
def classifica_inflacao(val):
    if val < 0:
        return 'Deflação'
    elif val <= 0.5:
        return 'Inflação Baixa'
    elif val <= 1.0:
        return 'Inflação Moderada'
    else:
        return 'Inflação Alta'

df_analise['classificacao'] = df_analise['valor'].apply(classifica_inflacao)

print("Contagem de meses por categoria de inflação:")
print(df_analise['classificacao'].value_counts())
```

Pergunta 9: Qual foi o ano com a maior média mensal de inflação?

```python
media_por_ano = df_analise.groupby('ano')['valor'].mean().reset_index()
maior_media_ano = media_por_ano.loc[media_por_ano['valor'].idxmax()]

print("Médias mensais de inflação por ano:")
print(media_por_ano)
print(f"\nO ano com a maior média mensal foi {int(maior_media_ano['ano'])}, com média de {maior_media_ano['valor']:.2f}% ao mês.")
```

Pergunta 10: Qual é a inflação geométrica acumulada total do período analisado?

```python
# Fórmula de acúmulo composto: Prod(1 + i/100) - 1
fator_acumulado = (1 + df_analise['valor'] / 100).prod() - 1
inflacao_total_acumulada = fator_acumulado * 100

print(f"Inflação total acumulada em todo o período analisado: {inflacao_total_acumulada:.2f}%")
```

Análise Comparativa por Governo

```python
import pandas as pd

# 1. Função para calcular a inflação acumulada composta (fórmula geométrica)
def calcula_acumulada(serie):
    fator = (1 + serie / 100).prod()
    return (fator - 1) * 100

# 2. Criando uma coluna para classificar o Governo com base na data
def define_governo(data):
    if data <= pd.to_datetime('2018-12-31'):
        return 'Governo Temer (2016-2018)'
    elif data <= pd.to_datetime('2022-12-31'):
        return 'Governo Bolsonaro (2019-2022)'
    else:
        return 'Governo Lula (2023-Atual)'

df_analise['governo'] = df_analise['data'].apply(define_governo)

# 3. Agrupando os dados por Governo e calculando métricas (Total de meses, Média mensal e Acumulado total)
resumo_governos = df_analise.groupby('governo')['valor'].agg(
    meses='count',
    media_mensal='mean',
    inflacao_acumulada=calcula_acumulada
).reset_index()

# Ordenando cronologicamente os governos
ordem_governos = ['Governo Temer (2016-2018)', 'Governo Bolsonaro (2019-2022)', 'Governo Lula (2023-Atual)']
resumo_governos['governo'] = pd.Categorical(resumo_governos['governo'], categories=ordem_governos, ordered=True)
resumo_governos = resumo_governos.sort_values('governo').reset_index(drop=True)

print("--- RESUMO COMPARATIVO DA INFLAÇÃO (INPC) POR GOVERNO ---")
print(resumo_governos.to_string(index=False))
```

Análise de Impacto: A Inflação (INPC) e a Perda de Poder de Compra por Gestão Presidencial

Para além do recorte macro temporal, é fundamental responder à pergunta central da nossa investigação: em qual período a classe trabalhadora sofreu a maior corrosão no seu poder de compra?

Como o INPC (Índice Nacional de Preços ao Consumidor) mede a variação do custo de vida para famílias com rendimento de 1 a 5 salários mínimos, ele é o termômetro mais fiel do impacto inflacionário sobre a base da pirâmide social. Utilizando o método de acumulação geométrica composta no Pandas (.prod()), que reflete o efeito de juros compostos da inflação mês a mês, podemos fatiar a série histórica nos seguintes mandatos:

Governo Michel Temer (Maio/2016 a Dezembro/2018)

Governo Jair Bolsonaro (Janeiro/2019 a Dezembro/2022)

Governo Luiz Inácio Lula da Silva (Janeiro/2023 até o período recente)

Discussão dos Resultados:

Ao observar os dados consolidados pelo agrupamento de governos (groupby), é possível traçar um panorama de como cada gestão lidou com as pressões inflacionárias globais e internas (como crises de commodities, choques energéticos, câmbio e políticas fiscais/monetárias):

O Marco Inicial (Governo Temer): Foi no pós-impeachment e durante a gestão Temer que se consolidou a guinada na política de indexação e o descolamento gradual do poder de compra frente aos índices de preços, marcando o início da série analisada com pressões significativas sobre os custos de habitação, alimentação e energia para as famílias de baixa renda.

O Choque Pandêmico e Pós-Pandemia (Governo Bolsonaro): Caracterizado por volatilidade extrema nos preços dos alimentos e combustíveis (impulsionados pela desvalorização cambial e gargalos globais nas cadeias de suprimento pós-crise sanitária), acumulando patamares elevados de inflação mensurados pelo INPC.

A Desaceleração Recente (Governo Lula): Representa o período de convergência gradual da inflação para as metas, com acomodação dos preços de itens essenciais da cesta básica nos últimos anos da série.



Referências e Contextualização Histórica
Governo Temer (2016-2018): Período marcado por reformas fiscais estruturais (como o teto de gastos) e forte impacto da crise econômica herdada, refletindo-se na rigidez dos custos de habitação e serviços essenciais mensurados pelo INPC.

Governo Bolsonaro (2019-2022): Forte pressão inflacionária impulsionada pela crise sanitária global da COVID-19, desvalorização cambial e choques nos preços internacionais de commodities, combustíveis e alimentos.

Governo Atual (2023-2026): Fase de reacomodação dos índices de preços e convergência gradual da inflação de alimentos e bens básicos.
